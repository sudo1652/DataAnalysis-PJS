# PJS 3-22

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('temp_pjs.csv', encoding='cp949')
df.head(3)
```

# PJS 3-23

```python
df.info()
```

# PJS 3-24

```python
df2 = df.fillna(method='ffill')
df2.info()
```

# PJS 3-25

```python
df2.rename(columns={'평균기온(°C)':'avg_temp'}, inplace=True)
df2.rename(columns={'최고기온(°C)':'max_temp'}, inplace=True)
df2.info()
```

# PJS 3-26

```python
df2['일시'] = pd.to_datetime(df2['일시'], format='%Y-%m-%d')
df2.info()
```

# PJS 3-27

```python
df3 = df2.set_index('일시')
df3.head(3)
```

# PJS 3-28

```python
df_Sep = df3[pd.DatetimeIndex(df3.index).month == 9]
df_Sep
```

# PJS 3-29

```python
plt.title('Seoul 2024. 9')
plt.plot(range(0, 30), df_Sep['max_temp'], c='r')
plt.plot(range(0, 30), df_Sep['avg_temp'], c='y')

plt.xlim(1, 30)
plt.ylim(20,37)
plt.legend()
plt.rcParams['figure.figsize'] = (8, 8)
plt.show()
```
